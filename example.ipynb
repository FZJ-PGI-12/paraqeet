## Example setup

```python
import numpy as np

from cthree.Optimiser import Optimiser
from cthree.generation.Generator import Generator
from cthree.measurement.UnitaryFidelity import UnitaryFidelity
from cthree.model.ClosedModel import ClosedModel
from cthree.model.Hamiltonian import Hamiltonian
from cthree.propagation.RungeKutta import RungeKutta
from cthree.Quantity import Quantity
```

```python
# The generator contains a list of devices which generate the drive fields
devices = []
generator = Generator(devices)

# The hamiltonian uses the generator and a list of subsystems, couplings and drive lines to create a matrix representation
subsystems = []
couplings = []
drives = []
hamiltonian = Hamiltonian(subsystems, couplings, drives, generator)

# The model determines the equation of motion, e.g. if it is an open or closed system
model = ClosedModel(hamiltonian)

# The propagation class solves the equation of motion
t_final = Quantity(value=20e-9, min=15e-9, max=25e-9, unit="s")
propagation = RungeKutta(model, t_final)

# An observable is measured after the propagation
idealGate = np.array([[0, 1], [1, 9]])
measurement = UnitaryFidelity(propagation, idealGate)

# The optimiser minimises the measured value and updates all parameters
optimisables = [generator, hamiltonian, propagation]
optimiser = Optimiser(measurement, optimisables)
```
